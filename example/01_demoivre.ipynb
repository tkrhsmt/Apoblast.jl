## Example 01: $D_n$-invariant polynomials

In this example, we will compute $D_n$-invariant polynomials, where $D_n$ is the dihedral group of order $2n$. The dihedral group $D_n$ can be generated by two transformations: a rotation and a reflection. We will use the following transformations:
- Rotation: $R_\theta(x, y) = (x \cos \theta + y \sin \theta, y \cos \theta - x \sin \theta)$
- Reflection: $S(x, y) = (x, -y)$

```julia
include("../src/Apoblast.jl")
using .Apoblast
using SymPy
```

```julia
model = Model(("x", "y"), ())

x, y = model.coords
```

```julia
function listorder(monoterms, order)
    isempty(monoterms) && return [1]

    return vcat([
        [(monoterms[1]^i) * t
         for t in listorder(monoterms[2:end], order - i)]
        for i in 0:order
    ]...)
end
list = listorder([x, y], 10)
lib = Library(model, Tuple(list))
```

```julia
trans_1 = Transformation(model,
    (x, -y),
    ();
)

θ = Sym(2) / Sym(5) * Sym(pi)
trans_2 = Transformation(model,
    (x*cos(θ) + y*sin(θ), y*cos(θ) - x*sin(θ)),
    ();
)
```

```julia
Li = [Sym(1)]
f̃ = [Sym(1)]

result = collect_follower(model, lib, Li, f̃, trans_1, trans_2)

for list in result
    display(list)
end
```
